```python
import re
import sys
import warnings

import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

warnings.filterwarnings("ignore")

TARGET = "Credit_Score"
DROP_COLS = ["ID", "Customer_ID", "Month", "Name", "SSN", "Occupation", "Type_of_Loan"]


# ----------------------------------------------------------------------------
# 1. Data loading
# ----------------------------------------------------------------------------
def make_synthetic_data(n=5000, seed=42):
    """Generate a realistic-looking demo dataset so the script runs out of the box."""
    rng = np.random.default_rng(seed)
    df = pd.DataFrame({
        "Age": rng.integers(18, 70, n),
        "Annual_Income": rng.normal(50000, 20000, n).clip(8000, 200000),
        "Num_Bank_Accounts": rng.integers(0, 10, n),
        "Num_Credit_Card": rng.integers(0, 10, n),
        "Interest_Rate": rng.integers(1, 35, n),
        "Num_of_Loan": rng.integers(0, 9, n),
        "Delay_from_due_date": rng.integers(0, 60, n),
        "Num_of_Delayed_Payment": rng.integers(0, 25, n),
        "Num_Credit_Inquiries": rng.integers(0, 20, n),
        "Credit_Mix": rng.choice(["Good", "Standard", "Bad"], n),
        "Outstanding_Debt": rng.gamma(2.0, 800, n),
        "Credit_Utilization_Ratio": rng.normal(32, 6, n).clip(15, 50),
        "Credit_History_Age": rng.integers(24, 400, n),
        "Payment_of_Min_Amount": rng.choice(["Yes", "No"], n),
        "Total_EMI_per_month": rng.gamma(2.0, 80, n),
        "Monthly_Balance": rng.normal(350, 150, n),
    })
    # Latent score -> class label (higher = better credit)
    score = (
        -0.04 * df["Delay_from_due_date"]
        - 0.10 * df["Num_of_Delayed_Payment"]
        - 0.0006 * df["Outstanding_Debt"]
        - 0.08 * df["Interest_Rate"]
        - 0.10 * df["Num_Credit_Inquiries"]
        + 0.004 * df["Credit_History_Age"]
        + df["Credit_Mix"].map({"Good": 1.5, "Standard": 0.0, "Bad": -1.5})
        + df["Payment_of_Min_Amount"].map({"Yes": -0.3, "No": 0.5})
        + rng.normal(0, 0.8, n)
    )
    df[TARGET] = pd.cut(score, [-np.inf, score.quantile(0.30), score.quantile(0.75), np.inf],
                        labels=["Poor", "Standard", "Good"]).astype(str)
    return df


def load_data(path=None):
    if path:
        print(f"Loading dataset from {path} ...")
        path = sys.argv[1] if len(sys.argv) > 1 else None
    print("No CSV supplied -> generating synthetic demo data.")
    return make_synthetic_data()


# ----------------------------------------------------------------------------
# 2. Cleaning
# ----------------------------------------------------------------------------
def parse_history_age(value):
    """'22 Years and 9 Months' -> 273 (months)."""
    if pd.isna(value):
        return np.nan
    if isinstance(value, (int, float)):
        return float(value)
    m = re.match(r"(\d+)\s*Years?\s*and\s*(\d+)\s*Months?", str(value))
    return int(m.group(1)) * 12 + int(m.group(2)) if m else np.nan


def clean_data(df):
    df = df.copy()
    df = df.drop(columns=[c for c in DROP_COLS if c in df.columns])

    # Credit_History_Age text -> months
    if "Credit_History_Age" in df.columns:
        df["Credit_History_Age"] = df["Credit_History_Age"].apply(parse_history_age)

    # Numeric columns polluted with stray characters ("34_", "-500", "_")
    for col in df.columns:
        if col == TARGET or df[col].dtype != object:
            continue
        cleaned = df[col].astype(str).str.replace(r"[_,]", "", regex=True)
        converted = pd.to_numeric(cleaned, errors="coerce")
        # Treat as numeric if most values converted successfully
        if converted.notna().mean() > 0.9:
            df[col] = converted

    # Placeholder junk values -> NaN in categoricals
    for col in df.select_dtypes(include="object").columns:
        if col != TARGET:
            df[col] = df[col].replace({"_": np.nan, "!@9#%8": np.nan, "NM": np.nan, "nan": np.nan})

    # Impossible values
    if "Age" in df.columns:
        df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = np.nan

    # Cap extreme outliers at 1st/99th percentile
    for col in df.select_dtypes(include=np.number).columns:
        lo, hi = df[col].quantile([0.01, 0.99])
        df[col] = df[col].clip(lo, hi)

    df = df.dropna(subset=[TARGET])
    return df


# ----------------------------------------------------------------------------
# 3. Model
# ----------------------------------------------------------------------------
def build_pipeline(X):
    num_cols = X.select_dtypes(include=np.number).columns.tolist()
    cat_cols = X.select_dtypes(exclude=np.number).columns.tolist()

    preprocess = ColumnTransformer([
        ("num", Pipeline([("impute", SimpleImputer(strategy="median")),
                          ("scale", StandardScaler())]), num_cols),
        ("cat", Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                          ("onehot", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
    ])

    model = RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=2,
        class_weight="balanced",
        n_jobs=-1,
        random_state=42,
    )
    return Pipeline([("prep", preprocess), ("clf", model)])


def main():
    path = sys.argv[1] if len(sys.argv) > 1 else None
    df = clean_data(load_data(path))
    print(f"Dataset shape after cleaning: {df.shape}")
    print("Class distribution:\n", df[TARGET].value_counts(normalize=True).round(3), "\n")

    X, y = df.drop(columns=[TARGET]), df[TARGET]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42)

    pipe = build_pipeline(X)
    pipe.fit(X_train, y_train)

    # Evaluation
    preds = pipe.predict(X_test)
    print(f"Test accuracy: {accuracy_score(y_test, preds):.4f}")
    cv = cross_val_score(pipe, X_train, y_train, cv=5, scoring="f1_macro", n_jobs=-1)
    print(f"5-fold CV macro-F1: {cv.mean():.4f} (+/- {cv.std():.4f})\n")
    print("Classification report:\n", classification_report(y_test, preds))
    labels = sorted(y.unique())
    print("Confusion matrix (rows=true, cols=pred):", labels)
    print(confusion_matrix(y_test, preds, labels=labels), "\n")

    # Feature importance
    names = pipe.named_steps["prep"].get_feature_names_out()
    imp = pd.Series(pipe.named_steps["clf"].feature_importances_, index=names)
    print("Top 10 features:\n", imp.sort_values(ascending=False).head(10).round(4), "\n")

    # Predict a single new customer (works with the synthetic columns)
    if path is None:
        sample = X_test.iloc[[0]]
        print("Sample customer prediction:", pipe.predict(sample)[0],
              "| actual:", y_test.iloc[0])


if __name__ == "__main__":
    main()
```
